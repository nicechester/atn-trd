## Integrating Gemini for Stock Analysis and FinBERT for Sentiment Decision

Now, let's use the Gemini API to generate an analysis for a hypothetical stock symbol. We will then pass this analysis to our FinBERT model to get a sentiment, which can inform a trading decision.

### Gemini API Setup

To use the Gemini API, ensure you have an API key stored in Colab secrets under the name `GOOGLE_API_KEY`. If you've already configured your Gemini API, you can skip the setup code block below.

```python
# Import the Python SDK for Gemini
import google.generativeai as genai
# Used to securely store your API key
from google.colab import userdata

# Configure the API key. You can skip this if already done.
try:
    GOOGLE_API_KEY=userdata.get('GOOGLE_API_KEY')
    genai.configure(api_key=GOOGLE_API_KEY)
    print("Gemini API configured successfully.")
except userdata.SecretNotFoundError:
    print("ERROR: GOOGLE_API_KEY not found in Colab secrets. Please add it to use Gemini API. (Click the '🔑' icon on the left panel).")
    raise
except Exception as e:
    print(f"ERROR configuring Gemini API: {e}. Please ensure your API key is valid and not leaked. If leaked, generate a new one from Google AI Studio and update it in Colab secrets.")
    raise
```

### Initialize Gemini Model

```python
# Initialize the Gemini model
gemini_model = genai.GenerativeModel('gemini-3.1-flash-lite')
```

```python
from transformers import pipeline

# Initialize FinBERT sentiment analyzer
sentiment_analyzer = pipeline('sentiment-analysis', model='ProsusAI/finbert')
print("FinBERT sentiment_analyzer initialized.")
```

### Get LLM Analysis and FinBERT Sentiment

```python
# Define the stock symbol you are interested in
stock_symbol = "GOOGL"

# Create a prompt for Gemini to analyze the stock
llm_prompt = f"Provide a brief fundamental and technical analysis of {stock_symbol} in a concise paragraph suitable for an investment overview."

print(f"Getting Gemini's analysis for {stock_symbol}...")

# Generate analysis using Gemini
try:
    gemini_response = gemini_model.generate_content(llm_prompt)
    llm_analysis_text = gemini_response.text
    print("\n--- Gemini's Analysis ---")
    print(llm_analysis_text)

    # Pass Gemini's analysis to FinBERT
    print("\n--- FinBERT Sentiment Analysis ---")
    finbert_results = sentiment_analyzer(llm_analysis_text)
    finbert_sentiment = finbert_results[0]['label']
    finbert_score = finbert_results[0]['score']

    print(f"FinBERT Sentiment: {finbert_sentiment}, Score: {finbert_score:.4f}")

    # Decide on an order based on FinBERT's sentiment
    print("\n--- Trading Decision ---")
    if finbert_sentiment == 'Positive':
        print(f"FinBERT suggests a positive outlook for {stock_symbol}. Consider a 'Buy' order.")
    elif finbert_sentiment == 'Negative':
        print(f"FinBERT suggests a negative outlook for {stock_symbol}. Consider a 'Sell' or 'Avoid' order.")
    else: # Neutral
        print(f"FinBERT suggests a neutral outlook for {stock_symbol}. Consider a 'Hold' order or further research.")

except Exception as e:
    print(f"An error occurred: {e}")
    print("Please ensure your Gemini API is correctly configured and try again.")
```

## Step 1: Simulate Data Gathering with Gemini

Since directly integrating multiple financial APIs (Finhub, Yahoo Finance, FRED, etc.) would require many API keys and complex setup, we will use Gemini to *simulate* the process of gathering and summarizing this diverse financial information for a given stock symbol. This will produce a comprehensive text block that we can then feed back into Gemini for structured extraction.

```python
import json

# Define the stock symbol (already defined as GOOGL, keeping for clarity)
# stock_symbol = "GOOGL"

# Prompt Gemini to generate a summary of financial data
data_gathering_prompt = f"Provide a detailed summary of recent financial news, key fundamentals (e.g., P/E, EPS, revenue growth, debt-to-equity), and relevant macroeconomic factors impacting {stock_symbol}. Assume the current date for 'recent' context. Be comprehensive but concise, suitable as input for further structured analysis."

print(f"\n--- Simulating data gathering for {stock_symbol} using Gemini ---")

try:
    # Generate the comprehensive financial data summary
    simulated_financial_data_response = gemini_model.generate_content(data_gathering_prompt)
    simulated_financial_data = simulated_financial_data_response.text
    print(simulated_financial_data)

except Exception as e:
    print(f"Error simulating data gathering with Gemini: {e}")
    simulated_financial_data = "Error: Could not generate simulated financial data."
```

## Step 2: Gemini for Structured Output

Now, we'll feed the `simulated_financial_data` into another Gemini prompt to generate a structured JSON output. This output will contain key metrics and a concise overall sentiment summary, making it easier for FinBERT to process.

```python
# Prompt Gemini to produce structured JSON output
structured_output_prompt = (f"Given the following financial analysis for {stock_symbol}, generate a JSON object with the following keys:\n"
                            "- 'stock_symbol': The ticker symbol.\n"
                            "- 'overall_sentiment_summary': A concise summary of the overall sentiment (positive, negative, or neutral) based on the analysis.\n"
                            "- 'key_takeaways': A list of 3-5 bullet points summarizing the most important aspects.\n"
                            "- 'recommendation_confidence_score': A score from 0 to 1 indicating the confidence in the implied recommendation (e.g., 0.8 for strong positive, 0.2 for weak negative).\n"
                            "- 'fundamental_highlights': A dictionary of key fundamental metrics and their values.\n"
                            "- 'technical_indicators_summary': A brief text summarizing technical outlook.\n\n"
                            f"Here is the financial analysis:\n{simulated_financial_data}\n\n"
                            "Ensure the output is valid JSON and directly parsable. Do not include any preambles or explanations outside the JSON block.")

print(f"\n--- Generating structured output for {stock_symbol} using Gemini ---")

try:
    structured_response = gemini_model.generate_content(structured_output_prompt)
    # Attempt to parse the response as JSON
    # Gemini might sometimes include extra text, so we try to find the JSON block.
    raw_json_string = structured_response.text.strip()
    # Find the first and last curly braces to isolate the JSON object
    first_brace = raw_json_string.find('{')
    last_brace = raw_json_string.rfind('}')
    if first_brace != -1 and last_brace != -1 and last_brace > first_brace:
        json_data_str = raw_json_string[first_brace : last_brace + 1]
    else:
        raise ValueError("Could not find a valid JSON object in Gemini's response.")

    structured_analysis = json.loads(json_data_str)
    print(json.dumps(structured_analysis, indent=2))

    # Extract the overall_sentiment_summary for FinBERT
    sentiment_text_for_finbert = structured_analysis.get('overall_sentiment_summary', '')
    if not sentiment_text_for_finbert:
        print("Warning: 'overall_sentiment_summary' not found in structured output. FinBERT will use an empty string.")

except json.JSONDecodeError as e:
    print(f"Error decoding JSON from Gemini's structured output: {e}")
    print(f"Raw response text: {structured_response.text}")
    sentiment_text_for_finbert = "Error: Could not decode structured sentiment."
except ValueError as e:
    print(f"Error processing Gemini's structured output: {e}")
    print(f"Raw response text: {structured_response.text}")
    sentiment_text_for_finbert = "Error: Could not extract structured sentiment."
except Exception as e:
    print(f"An unexpected error occurred during structured output generation: {e}")
    sentiment_text_for_finbert = "Error: Could not generate structured sentiment."
```

## Step 3: FinBERT Decision Making for Order Quantity

Now, we'll take the `overall_sentiment_summary` from Gemini's structured output and feed it into FinBERT. Based on FinBERT's sentiment and score, we'll define a logic to decide how much to buy, sell, or hold.

```python
from transformers import pipeline

# Ensure sentiment_analyzer is initialized (it should be from the new dedicated FinBERT setup cell)
# The try-except block here is commented out as sentiment_analyzer will now be initialized earlier.
# try:
#     _ = sentiment_analyzer # Check if it's already defined
# except NameError:
#     sentiment_analyzer = pipeline('sentiment-analysis', model='ProsusAI/finbert')
#     print("FinBERT sentiment_analyzer re-initialized.")


print("\n--- FinBERT Sentiment Analysis on Gemini's Summary ---")

if sentiment_text_for_finbert and not sentiment_text_for_finbert.startswith("Error:"):
    finbert_results_summary = sentiment_analyzer(sentiment_text_for_finbert)
    finbert_sentiment_summary = finbert_results_summary[0]['label'].lower() # Convert to lowercase for easier comparison
    finbert_score_summary = finbert_results_summary[0]['score']

    print(f"FinBERT Sentiment (from Gemini summary): {finbert_sentiment_summary.capitalize()}, Score: {finbert_score_summary:.4f}")

    # --- Placeholder for User's Budget and Position ---
    # In a real scenario, these would come from your brokerage API or internal records.
    account_balance = 50000.0  # Example: $50,000 available cash
    current_position_shares = 100 # Example: You currently own 100 shares of GOOGL
    current_stock_price = 175.0  # Example: Current price of GOOGL

    print(f"\n--- Current Portfolio Snapshot ---")
    print(f"Account Balance: ${account_balance:,.2f}")
    print(f"Current {stock_symbol} Position: {current_position_shares} shares @ ${current_stock_price:,.2f} = ${current_position_shares * current_stock_price:,.2f}")

    # --- Decision Logic for Order Quantity (Dynamic) ---
    def get_trading_recommendation_dynamic(sentiment, score, confidence_score, balance, current_shares, stock_price):
        action = "Hold"
        quantity = 0

        # Risk management: Define a max percentage of balance to use for a single trade
        max_allocation_percent = 0.05 # e.g., max 5% of balance for a strong buy/sell
        max_investment_amount = balance * max_allocation_percent

        if sentiment == 'positive':
            if score >= 0.9 and confidence_score >= 0.7: # Strong Buy
                # Buy up to max_investment_amount, but at least 1 share
                quantity = int(max(1, max_investment_amount / stock_price))
                action = "Strong Buy"
            elif score >= 0.7 and confidence_score >= 0.5: # Moderate Buy
                quantity = int(max(1, (max_investment_amount * 0.5) / stock_price)) # Half of max allocation
                action = "Buy"
            else: # Weak Buy
                quantity = int(max(0, (max_investment_amount * 0.2) / stock_price)) # Smaller allocation
                action = "Weak Buy"

            # Ensure we don't overspend
            if quantity * stock_price > balance:
                quantity = int(balance / stock_price)
                if quantity == 0 and action != "Hold": # If cannot even buy 1 share, then hold
                    action = "Hold"

        elif sentiment == 'negative':
            if score >= 0.9 and confidence_score >= 0.7: # Strong Sell
                quantity = min(current_shares, int(max(1, max_investment_amount / stock_price))) # Sell up to max allocation or current shares
                action = "Strong Sell"
            elif score >= 0.7 and confidence_score >= 0.5: # Moderate Sell
                quantity = min(current_shares, int(max(1, (max_investment_amount * 0.5) / stock_price))) # Sell half of max allocation
                action = "Sell"
            else: # Weak Sell
                quantity = min(current_shares, int(max(0, (max_investment_amount * 0.2) / stock_price))) # Smaller sell
                action = "Weak Sell"

            # If we don't have enough shares to sell any quantity, then hold
            if quantity == 0 and action != "Hold":
                action = "Hold"

        return action, quantity

    # Use the confidence score from Gemini's structured output if available, else default.
    gemini_confidence = structured_analysis.get('recommendation_confidence_score', 0.5) if 'structured_analysis' in locals() else 0.5

    action, quantity = get_trading_recommendation_dynamic(finbert_sentiment_summary, finbert_score_summary, gemini_confidence,
                                                           account_balance, current_position_shares, current_stock_price)

    print(f"\n--- Combined Trading Decision for {stock_symbol} (Dynamic Quantity) ---")
    print(f"Recommended Action: {action}")
    print(f"Suggested Quantity: {quantity}")

    if action.endswith('Buy') and quantity > 0:
        print(f"Execute a {action.upper()} order for {quantity} shares of {stock_symbol} (Estimated cost: ${quantity * current_stock_price:,.2f}).")
    elif action.endswith('Sell') and quantity > 0:
        print(f"Execute a {action.upper()} order for {quantity} shares of {stock_symbol} (Estimated proceeds: ${quantity * current_stock_price:,.2f}).")
    else:
        print(f"HOLD position for {stock_symbol}. No action taken based on current sentiment and portfolio constraints.")

else:
    print("Could not perform FinBERT analysis due to missing or errored sentiment summary.")
```
